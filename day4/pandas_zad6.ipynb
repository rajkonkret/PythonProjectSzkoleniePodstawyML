1. Wczytanie pliku
Załaduj dane z pliku (np. CSV) do obiektu DataFrame, aby móc je analizować i przetwarzać.
2. Zweryfikowanie kolumny z datą
Sprawdź, czy kolumna zawierająca daty ma poprawny format.
Jeśli nie, przekonwertuj ją na typ datetime, aby umożliwić sortowanie, filtrowanie i operacje czasowe.
3. Konwersja kolumny na typ datetime
Zamień kolumnę z datą na typ datetime, aby zapewnić poprawną interpretację wartości czasowych.
4. Zweryfikowanie kolumny volume
Upewnij się, że kolumna volume zawiera wartości liczbowe.
Jeśli znajdują się tam wartości tekstowe, puste lub błędne — oczyść je lub zamień na wartości możliwe do konwersji.
5. Konwersja kolumny volume na typ int
Po weryfikacji przekształć kolumnę volume na typ całkowity (int), aby umożliwić obliczenia i sortowanie.
6. Znalezienie 7 wierszy z najniższą wartością volume
Posortuj dane rosnąco według kolumny volume i wybierz pierwsze 7 rekordów, aby uzyskać obserwacje o najmniejszej wartości wolumenu.


```python
import pandas as pd

appl_df = pd.read_csv('AAPL.csv')
appl_df.info()
```

```python
appl_df['Date'] = pd.to_datetime(
    appl_df['Date'],
    errors="coerce"
)
appl_df.dtypes
```

```python
appl_df['Volume'].map(type).eq(str)
```

```python
appl_df['Volume'].map(type).eq(str).any()
```

```python
appl_df['Volume'].isna().sum()
```

```python
appl_df['Volume'] = appl_df["Volume"].astype(int)
```

```python
appl_df["Volume"] = pd.to_numeric(
    appl_df['Volume'],
    errors="coerce"
)
```

```python
appl_df['Volume'] = appl_df['Volume'].fillna(0)
```

```python
appl_df['Volume'] = appl_df["Volume"].astype(int)
```

```python
appl_df.info()
```

```python
result = appl_df.sort_values(by="Volume").head(7)
```

```python
result
```

```python
lowest_volume_7 = appl_df.nsmallest(7, "Volume")  # wyswietla najniższe wartości
```

```python
lowest_volume_7
```
