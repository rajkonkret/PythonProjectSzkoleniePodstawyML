```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
# loc - średnia wartość
# scale - odch standardowe
# size - liczba elementów

dane = np.random.normal(loc=70, scale=10, size=30)
```

```python
df = pd.DataFrame({"Wynik": dane})
df.head()
```

```python
df.describe()
```

```python
# wariancja - im wyzsza tym dane sa bardziej rozproszeno wokół średniej
wariancja = df['Wynik'].var()
```

```python
wariancja
```

```python
x = np.random.normal(50, 10, 100)
y = 2 * x + np.random.normal(0, 10, 100)
```

```python
df = pd.DataFrame({"x": x, "y": y})
```

```python
sns.regplot(x="x", y="y", data=df, color="green")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data=tips, x="total_bill", kde=True)
plt.title("Rozkład rachunków")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Rachunki wg dnia tygodnia")
plt.show()
```

```python
# regresja: napiwek vs rachunek
plt.figure(figsize=(6, 4))
sns.regplot(data=tips, x="total_bill", y="tip", scatter=True)
plt.title("Rachunek vs Napiwek")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

df = sns.load_dataset("iris")

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Macierz korelacji")
plt.show()
```
