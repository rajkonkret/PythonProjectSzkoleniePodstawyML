```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
# loc - średnia wartość
# scale - odch standardowe
# size - liczba elementów

dane = np.random.normal(loc=70, scale=10, size=30)
```

```python
df = pd.DataFrame({"Wynik": dane})
df.head()
```

```python
df.describe()
```

```python
# wariancja - im wyzsza tym dane sa bardziej rozproszeno wokół średniej
wariancja = df['Wynik'].var()
```

```python
wariancja
```

```python
import seaborn as sns
```

```python
from matplotlib import pyplot as plt

plt.figure(figsize=(12, 5))
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df['Wynik'], kde=True, color="skyblue")
plt.title("Histogram")
```

### Regresja liniowa

```python
x = np.random.normal(50, 10, 100)
y = 2 * x + np.random.normal(0, 10, 100)
```

```python
df = pd.DataFrame({"x": x, "y": y})
```

```python
sns.regplot(x="x", y="y", data=df, color="green")
```

```python
# wykresy interaktywne - plotly
import plotly.express as px
```

```python
kategoria = ['A', 'B', "C", "D"]
wartosci = [10, 23, 17, 5]
```

```python
fig = px.bar(
    x=kategoria,
    y=wartosci,
    labels={"x": "Kategorie", "y": "Wartości"},
    title="Interaktywny wykres słupkowy (Plotly)"
)
fig.show()
```
