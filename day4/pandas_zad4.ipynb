```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('orders.csv')
df
```

```python
df.info()
```

```python
# usunięcie kolumny
df = df.drop(columns=["total"])
df
```

```python
# dodanie nowej kloumny, nowy parametr
df['total'] = df['price'] * df['quantity']
df
```

```python
# wyszukac wiersze z "Warszawa"
df.loc[df['city'] == "Warszawa"]
```

```python
# total > 4000
df.loc[df['total'] > 4000]
```

```python
# wypisac dla Krakowa -> customer, kwota
df.loc[df['city'] == "Kraków", ['customer', 'city', 'price']]
```

```python
# zapłacone, Laptop
df.loc[(df['category'] == "Laptop") & (df['paid'] == True),
['customer', 'category', 'total', 'paid']]
```

```python
df.loc[(df['category'] == "Laptop") & (df['paid']),
['customer', 'category', 'total', 'paid']]
```

```python
# Warszawa i Gdańsk
df.loc[df['city'].isin(['Warszawa', "Gdańsk"])]
```

```python
# wszystkie inne not ~
df.loc[~df['city'].isin(['Warszawa', "Gdańsk"])]
```

```python
# # nan w discount
df.loc[df['discount'].isna()]
```

```python
df.isna().sum()
```

```python
# notna()
df.loc[df['discount'].notna()]
```

```python
# zamiana NaN na 0
df.loc[df['discount'].isna(), "discount"] = 0
```

```python
df
```

```python
df['discount'] = df['discount'].fillna(0)
```

```python
# w pola status wpisz "zaplacone" tam gdzie jest True
df.loc[df['paid'] == True, 'paid'] = "zaplacone"
```

```python
df.loc[df['paid'], 'status'] = "zapłacone"
```

```python
df
```

```python
# niezapłacone
df.loc[df['paid'] == False, 'status'] = 'niezapłacone'
df
```

```python
df['paid'].dtype
```

```python
df.loc[~df['paid'], 'status'] = 'niezapłacone'  # może nie działać w Pandas 2
```

```python
df
```

```python
# dla zamówien powyżej 5000 wpisz w kolumnie 'segment' wartośc 'VIP'
df.loc[df['total'] > 5000, "segment"] = "VIP"
```

```python
df.loc[df['total'] <= 5000, "segment"] = "normal"
```

```python
df
```

```python
df['total'].dtype
```

```python
# zamiana typu kolumny
df['total'] = df['total'].astype('int64')
```

```python
# wypełnienie w jednym kroku spełniających i niespełniających
# za pomocą np.where
df['segment'] = np.where(df['total'] > 5000, "VIP", "normal")
df
```
