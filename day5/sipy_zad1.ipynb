```python
# pole pod wykresem
import numpy as np
from scipy import integrate
```

```python
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 4, 6, 8])
```

```python
pole = integrate.trapezoid(y, x)
pole
```

```python
# A * x = b

from scipy import linalg

A = np.array([[3, 2], [1, 4]])
b = np.array([6, 5])

x = linalg.solve(A, b)
print(x)
```
