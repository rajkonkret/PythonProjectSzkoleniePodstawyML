## Wskazówka – pandas 3

W pandas 3 obowiązuje mechanizm Copy-on-Write.

Unikaj przypisywania łańcuchowego, np.:

`df["price"].fillna(srednia, inplace=True)`

Zamiast tego przypisz wynik do kolumny:

`df["price"] = df["price"].fillna(srednia)`

Możesz również użyć `DataFrame.fillna()` z parametrem `inplace=True`, przekazując słownik wartości dla wybranych kolumn.



```python
import pandas as pd
```

```python
df = pd.read_csv("orders_example.csv")
df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.head(8)
```

```python
df.isnull().sum()
```

```python
df['price'].fillna(df['price'].mean(), inplace=True)  # działa w PAndas 2, nie działa w Pandas 3
```

```python
# dla Pandasa 3
df['price'] = df['price'].fillna(df['price'].mean())
```

```python
df['price']
```

```python
df.info()  #  4   price        6 non-null      float64
```

```python
df['score'] = df['score'].fillna(df['score'].median())
```

```python
df.isnull().sum()
```

# Wartości odstające

```python
import pandas as pd
```

```python
# metoda kwantylowa
df = pd.read_csv('orders_example.csv')
df
```

```python
df.describe()
```

```python
Q1 = df['score'].quantile(0.25)
Q3 = df['score'].quantile(0.75)
```

```python
print(Q1, Q3)
```

```python
IQR = Q3 - Q1
IQR
```

```python
outliers = df[df['score'] > Q3 + 1.5 * IQR]
```

```python
outliers
```

```python
median_score = df['score'].median()
```

```python
df.loc[df['score'] > Q3 + 1.5 * IQR, "score"] = median_score
```

```python
df
```

```python
# posortowac wg order_date
# score > 85 dla Poland
```

```python
df.sort_values(by="order_date")
```

```python
# score > 85 dla Poland
df[df.score > 85]
```

```python
df[df['score'] > 85]
```

```python
df[df['country'] == "Poland"]
```

```python
df[(df['score'] > 85) & (df['country'] == "Poland")]
```

```python
# query
df.query("country == 'Poland' and score > 85")
```

```python
# electronics, price > 1000
df.query("category == 'Electronics' and price > 1000")
```

```python
# policzyc ile zamówień jest w danej kategorii
# pogrupowac  wg category, order_id
df.groupby('category').agg({"order_id": "count"})
```

```python
# order_id, score -> mean
df.groupby('category').agg({"order_id": "count", "score": "mean"})
```

```python
df.groupby('category').agg({"order_id": "count", "score": ["mean", "sum"]})
```

```python
df.groupby('category').agg({"score" : "mean"})
```
