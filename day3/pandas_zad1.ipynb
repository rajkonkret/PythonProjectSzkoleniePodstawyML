```python
movies_df = pd.read_csv("IMDB-Movie-Data-with-duplicates-and-nulls.csv", index_col="Title")
movies_df.head()
```

```python
movies_df.info()
```

```python
movies_df.describe()
```

```python
movies_df.duplicated().any()  # czy są jakiekolwiek duplikaty, cały wiersz
```

```python
movies_df.duplicated().sum()  # ile jest duplikatów
```

```python
movies_df["Rank"].duplicated().any()
```

```python
movies_df["Rank"].duplicated().sum()
```

```python
movies_df.duplicated(subset=["Rank"]).any()
```

```python
movies_df.duplicated()
```

```python
movies_df[movies_df.duplicated() == True]  # wyświetl duplikaty
```

```python
movies_df[movies_df.duplicated(keep=False)]  # pokaze duplikat i oryginał
```

```python
movies_df_not_dupl = movies_df.drop_duplicates()  # usunięcie duplikatów
movies_df_not_dupl.head(2)
```

```python
# sprawdzić czy są nadal duplikaty
movies_df_not_dupl.duplicated().any()
```
